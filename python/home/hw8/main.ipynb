import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

```python
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print("Оригінальна структура Series:\n", s)

print("\nЗначення за індексом 'c':", s['c'])
print("Зріз (перші 3 елементи):\n", s[:3])

print("\nФільтрація (значення > 25):\n", s[s > 25])

print("\nМноження на 2:\n", s * 2)
```

```python
data = {
    'Місяць': ['Січ', 'Лют', 'Бер', 'Кві', 'Тра'],
    'Дохід': [1500, 2200, 1800, 2500, 2100],
    'Витрати': [1000, 1100, 1200, 1300, 1150]
}
df = pd.DataFrame(data)
print("Оригінальний DataFrame:\n", df)

df['Прибуток'] = df['Дохід'] - df['Витрати']
print("\nDataFrame після додавання стовпця 'Прибуток':\n", df)

print("\nТільки стовпець 'Прибуток':\n", df['Прибуток'])
print("\nМісяці з прибутком більше 800:\n", df[df['Прибуток'] > 800])

print("\nОписова статистика:\n", df.describe())
```

```python
x = np.linspace(0, 10, 100)
df_plot = pd.DataFrame({
    'X': x,
    'Синус (Sin)': np.sin(x),
    'Косинус (Cos)': np.cos(x),
    'Лінійна (Linear)': x * 0.5 - 2
})

def plot_data(column, color, show_grid):
    plt.figure(figsize=(10, 5))
    plt.plot(df_plot['X'], df_plot[column], color=color, linewidth=2.5)
    
    plt.title(f'Графік функції: {column}')
    plt.xlabel('Ось X')
    plt.ylabel('Значення Y')
    
    if show_grid:
        plt.grid(True, linestyle='--', alpha=0.7)
    
    plt.show()
widgets.interact(
    plot_data,
    column=widgets.Dropdown(
        options=['Синус (Sin)', 'Косинус (Cos)', 'Лінійна (Linear)'], 
        value='Синус (Sin)', 
        description='Функція:'
    ),
    color=widgets.ColorPicker(
        concise=False, 
        description='Колір:', 
        value='blue'
    ),
    show_grid=widgets.Checkbox(
        value=True, 
        description='Показувати сітку'
    )
);
```
